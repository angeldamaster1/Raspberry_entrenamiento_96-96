#Crear modelo de keras a .tflite

Ten en cuenta de que ya debes de tener tu modelo ya entrenado para que lo puedas utilizar en la parte de : *from_keras_model(TuModeloAqui)*

```python
# Convert the model.
converter = tf.lite.TFLiteConverter.from_keras_model(TuModeloAqui)
tflite_model = converter.convert()

# Save the model.
with open('NombreDeEjemplo.tflite', 'wb') as f:
  f.write(tflite_model)
```
